# Train the Sahpaathi perception model

Turns learners' own labelled moments into `weights.json` for the AI teacher's perception agents.

**You need:** `train_sense.py` (from the repo's `ml/` folder) and one or more CSVs downloaded from the AI teacher (the download button next to *I get it / I'm lost*). The CSVs hold numbers only: no pictures, no sound.

**Do not** use `synthetic.csv` for real weights: it is invented, to test the pipeline.

```
import json
w = json.load(open('weights.json'))
print(w['metrics'])
sorted(w['models']['confused']['coef'].items(), key=lambda kv: -abs(kv[1]))[:8]
```

If the cross-validated **AUC is below about 0.7**, the model barely beats guessing: collect more labelled moments (different people, rooms and lighting) before using it.

Otherwise download `weights.json` and put it at `backend/app/tutor/sense/weights.json`, then open a pull request. The server picks it up on the next start.
